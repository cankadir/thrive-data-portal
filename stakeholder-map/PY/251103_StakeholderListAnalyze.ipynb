```python
import pandas as pd
import os
pd.set_option( "display.max_columns", None )

# path = r"C:\Git_Repo\thrive-data-portal\stakeholder-map\data"
# file_path = os.path.join( path , "260326_StakeholderList.xlsx")

file_path = "https://docs.google.com/spreadsheets/d/e/2PACX-1vS-Br0k_jjYJTYkqp26z8Ujg-ydljDdzS0qB-By3Fj_vV4-FOhTk7-wnFzuqrNMC8q2Y1Hnk5XF2I5m/pub?gid=936337613&single=true&output=tsv"
sl = pd.read_csv( file_path , delimiter="\t")
sl = sl.dropna( axis=1, how="all")
sl.head()
```

```python
sl = sl.drop( columns = list(sl.columns[ sl.columns.str.contains("Unnamed")]))
sl.columns = [i.strip() for i in sl.columns]
sl.head()
```

```python
cols = sl.columns

for col in cols:
    if sl[col].dtype == 'obj':
        sl[col] = sl[col].str.strip()

sl.head()
```

```python
sl['Sector_split'] = sl['Sector'].str.replace(" ", '').str.strip().str.split(",")
sl.explode("Sector_split").head(3)
```

```python
sl['counties_split'] = sl['county(s) of service'].str.replace(" ", '').str.strip().str.split(",")
sl
```

```python
sl['Organization'] = sl['Organization'].str.upper().str.strip()

orgs_list = sl[ ~sl['Organization'].isin(["THRIVE REGIONAL PARTNERSHIP",'Studio 2263']) ] \
    .groupby(['Sector','county(s) of service',"Organization"]).agg({
        "First Name": len,
        "Tags": list
    })#.size().reset_index().set_index(['Sector','Region',"Organization"]).sort_values(by=0, ascending=False)

orgs_list.columns = ['Count','Tags']
orgs_list.sort_values(by="Count", ascending=False)
```

```python
orgs_list.to_excel( r"C:\Git_Repo\thrive-data-portal\stakeholder-map\data\260326_UniqueOrgs.xlsx" )
```

```python
import pandas as pd
from fuzzywuzzy import fuzz
from fuzzywuzzy import process

# Get unique organization names
org_names = sl['Organization'].dropna().unique()

# Find pairs of similar names (threshold can be adjusted)
similar_pairs = []
threshold = 85  # similarity score out of 100

for i, name in enumerate(org_names):
    for other in org_names[i+1:]:
        score = fuzz.token_sort_ratio(name, other)
        if score >= threshold:
            similar_pairs.append((name, other, score))

similar_pairs = pd.DataFrame( similar_pairs )
similar_pairs.columns = ['original','potential','score']
similar_pairs[similar_pairs['score'] > 85].sort_values(by='score')
```

## Merge Geographies

```python
import geopandas as gpd
import matplotlib.pyplot as plt

counties = gpd.read_file( r"C:\Users\cansu\Dropbox\studio2263\Thrive\GIS\upload_online_mapping\thrive_counties_boundaries.geojson")
counties = counties[['county_nam','geometry']]
counties
```

```python
by_county = sl.copy()

by_county = by_county.explode("counties_split")
by_county['counties_split'] = by_county['counties_split'].str.strip().str.upper()
by_county = by_county.dropna( subset=['counties_split','Organization'] , axis=0 )
by_county = by_county.groupby('counties_split').agg({
    'Organization': list,
    'Sector_split': list
})

by_county['Organization'] = by_county['Organization'].apply( lambda x: list(set([i.upper().strip() for i in x]) ) )
by_county['count'] = by_county['Organization'].apply( lambda x: len(x) )
by_county['Sector_split_2'] = by_county['Sector_split'].apply( lambda x: list(set([i[0]  for i in x if not pd.isna(i) ])) )
by_county.sort_values(by='count', ascending=False ).head(4)
```

```python
counties['county_nam'] = counties['county_nam'].str.upper().str.strip()
counties.head(2)
```

```python
by_county_map = by_county.join( counties.set_index('county_nam') ).dropna(subset=['geometry'])
by_county_map = gpd.GeoDataFrame( by_county_map.drop(columns='geometry') , geometry= by_county_map['geometry'] , crs=4326 )

fig,ax = plt.subplots()
by_county_map.plot( ax=ax, column='count', cmap="OrRd", legend=True, ec='w' )
ax.axis('off');
```
